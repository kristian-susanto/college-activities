***Menghitung persentase nilai hilang di setiap kolom DataFrame `df_univ` dan kemudian menghapus kolom-kolom yang memiliki persentase nilai hilang lebih dari 30%. Ini merupakan langkah pembersihan data yang penting untuk memastikan bahwa analisis yang dilakukan di masa depan lebih akurat dan tidak dipengaruhi oleh data yang tidak lengkap. Setelah menjalankan kode ini, `df_cleaned` akan berisi kolom-kolom yang lebih bersih, meningkatkan kualitas data yang digunakan untuk analisis selanjutnya.***

```python
# menghitung jumlah missing values
missing_count = df_cleaned.isnull().sum()

# menghitung persentase missing values
missing_percentage = df_cleaned.isnull().sum() / len(df_cleaned) * 100

# membuat DataFrame baru untuk menampilkan jumlah dan persentase berdampingan
missing_data = pd.DataFrame({
    'Missing Values': missing_count,
    'Percentage (%)': missing_percentage
})

# menampilkan hasil
missing_data
```

***Menunjukkan bahwa kolom yang sebelumnya memiliki missing value lebih dari 30% telah hilang, sekarang hanya menampilkan beberapa kolom yang memiliki missing value kurang dari 30%, untuk penanganan selanjutnya yang lebih baik ***

```python
# mengisi missing values untuk kolom numerik berdasarkan unit_measure
for column in df_cleaned.select_dtypes(include=['number']).columns:
    df_cleaned[column] = df_cleaned[column].transform(lambda x: x.fillna(x.median()))

# mengisi missing values dengan modus untuk kolom kategorikal
for column in df_cleaned.select_dtypes(include=['object']).columns:
    df_cleaned[column] = df_cleaned[column].fillna(df_cleaned[column].mode()[0])
```

***Mengisi nilai hilang pada kolom numerik dengan median karena lebih tahan terhadap outlier dan memberikan representasi yang lebih akurat untuk distribusi data yang tidak normal. Sementara itu, menggunakan modus untuk kolom kategorikal memastikan bahwa nilai yang paling umum tetap dipertahankan, meningkatkan konsistensi data.***

```python
# menghitung jumlah missing values
missing_count = df_cleaned.isnull().sum()

# menghitung persentase missing values
missing_percentage = df_cleaned.isnull().sum() / len(df_cleaned) * 100

# membuat DataFrame baru untuk menampilkan jumlah dan persentase berdampingan
missing_data = pd.DataFrame({
    'Missing Values': missing_count,
    'Percentage (%)': missing_percentage
})

# menampilkan hasil
missing_data
```

### Z-Score

```python
# menghitung Z-Score
data_zscore=df_obs[(np.abs(stats.zscore(df_obs)) < 3).all(axis=1)]
```

***Menghitung Z-Score untuk setiap nilai dalam DataFrame `df_obs` dan menyaring outlier dengan menghapus semua baris yang memiliki nilai Z-Score lebih besar dari 3 dalam kolom 'Production-based CO2 emissions', yang mengindikasikan bahwa nilai tersebut lebih dari tiga standar deviasi dari rata-rata.***

```python
fig, ax = plt.subplots(figsize=(3,6)) #buat frame/figure
sns.boxplot(data=data_zscore)
```

***Menunjukkan distribusi data untuk variabel "Production-based CO2 emissions" setelah menghilangkan outlier, dengan beberapa data tetap terdeteksi di luar batas atas, mengindikasikan variabilitas yang cukup besar dalam emisi CO2.***

### IQR

```python
#cari Q1, Q3, dan interquartile range
Q1 = df_obs.quantile(q=.25)
Q3 = df_obs.quantile(q=.75)
IQR = df_obs.apply(stats.iqr) #IQR=Q3-Q1 atau bisa menggunakan stats.iqr dari scipy

#ambil nilai antara 1.5*IQR dari Q1 dan Q3 aja
data_iqr = df_obs[~((df_obs< (Q1-1.5*IQR)) | (df_obs > (Q3+1.5*IQR))).any(axis=1)]
```

***Data yang difilter hanya menyisakan nilai-nilai yang berada dalam batas normal (1.5 * IQR dari Q1 dan Q3), sehingga outlier yang berada di luar batas tersebut tidak termasuk dalam hasil akhir.***

```python
# membuat frame/figure menggunakan matplotlib dengan ukuran (3,6) dan menampilkan boxplot dari data yang telah dibersihkan dari outlier berdasarkan Z-score
fig, ax = plt.subplots(figsize=(3,6)) #buat frame/figure
sns.boxplot(data=data_iqr)
```

***Plot ini menunjukkan distribusi "Production-based CO2 emissions" setelah menghapus outlier berdasarkan IQR, dengan tampilan yang lebih fokus pada data di dalam rentang normal.***

## 5. Feature Encoding

# **C.3 Feature Engineering**

## 1. Mengidentifikasi Tren Production-based CO2 emissions per tahun di setiap negara

```python
# mengelompokkan data berdasarkan negara dan tahun, kemudian menghitung rata-rata emisi CO2
# menyertakan 'development_status' dalam pengelompokan agar tetap ada dalam DataFrame yang dihasilkan
df_trend = df_cleaned.groupby(['nation', 'development_status', df_cleaned.index])['Production-based CO2 emissions'].mean().reset_index()

# mengganti nama kolom indeks menjadi 'period' untuk kejelasan
df_trend.rename(columns={df_trend.columns[2]: 'period'}, inplace=True) # Rename the third column to 'period'

# menghitung perubahan emisi CO2 tahun ke tahun untuk setiap negara
df_trend['CO2_Change'] = df_trend.groupby('nation')['Production-based CO2 emissions'].diff()

# membuat kolom baru untuk mengkategorikan tren
df_trend['CO2_Trend'] = 'Stable'
df_trend.loc[df_trend['CO2_Change'] > 0, 'CO2_Trend'] = 'Increasing'
df_trend.loc[df_trend['CO2_Change'] < 0, 'CO2_Trend'] = 'Decreasing'
```

***Feature ini dibuat untuk menganalisis perubahan emisi CO2 dari waktu ke waktu berdasarkan negara dan status pembangunan, guna memahami tren emisi yang dapat mempengaruhi kebijakan lingkungan dan strategi keberlanjutan di masing-masing negara, serta untuk mengidentifikasi pola peningkatan atau penurunan emisi yang mungkin berhubungan dengan faktor-faktor pembangunan ekonomi dan teknologi.***

```python
# menampilkan 12 data teratas dari df_trend yang hanya mencakup negara dengan development_status = 'Developed', disusun berdasarkan periode dari tiga tahun terakhir
df_trend[(df_trend['development_status'] == 'Developed')].sort_values(by=['period'], ascending=False).head(12)
```

***Data menunjukkan bahwa emisi CO2 produksi di negara-negara maju seperti Jepang, Korea, dan Singapura mengalami tren peningkatan pada tahun 2021, meskipun Hong Kong menunjukkan penurunan kecil dalam emisi selama periode yang sama, mencerminkan perbedaan dalam kebijakan dan praktik lingkungan di antara negara-negara tersebut.***


* **Variabel Target**: Variabel target yang dianalisis adalah CO2_Emissions_per_Unit_of_Energy_Supply, yang mengukur emisi karbon dioksida per unit pasokan energi.

* **Fitur Relevan**: Lima fitur dengan korelasi tertinggi yang dipilih untuk analisis lebih lanjut menunjukkan hubungan signifikan dengan emisi CO2.

* **Emisi CO2 Berbasis Produksi**: Fitur Production-based CO2 emissions menunjukkan bahwa emisi karbon dari produksi berperan penting dalam menentukan total emisi.

* **Pasokan Energi Total**: Total energy supply sebagai fitur mencerminkan hubungan antara total pasokan energi dan dampaknya terhadap emisi CO2.

* **Emisi Gas Rumah Kaca Berdasarkan Permintaan**: Fitur Demand-based GHG emissions menunjukkan bahwa emisi gas rumah kaca yang dihasilkan berdasarkan permintaan juga relevan terhadap emisi CO2.

* **Emisi Gas Rumah Kaca Berbasis Produksi**: Production-based GHG emissions mengindikasikan bahwa produksi gas rumah kaca juga berkontribusi pada emisi CO2.

* **Pasokan Energi Terbarukan**: Fitur Renewable energy supply menunjukkan potensi untuk mengurangi emisi CO2 melalui peningkatan penggunaan sumber energi terbarukan.

```python
# inisialisasi StandardScaler
scaler = StandardScaler()

# fit dan transformasi data fitur X (scaling fitur numerik)
X_scaled = scaler.fit_transform(X)

# jika ingin mengonversi kembali ke DataFrame dengan mempertahankan nama kolom:
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# output hasil scaling
print("Fitur numerik telah di-scale menggunakan StandardScaler:")
X_scaled_df.head()
```


* **Transformasi Data**: Fitur numerik dalam dataset telah di-scale, yang bertujuan untuk menjadikan semua fitur berada dalam skala yang sama.

* **Standardisasi**: Proses scaling ini menggunakan standardisasi, di mana nilai fitur dikurangi dengan rata-rata dan dibagi dengan deviasi standar, menghasilkan nilai dengan rata-rata 0 dan deviasi standar 1.

* **Pentingnya Scaling**: Scaling fitur numerik sangat penting untuk model machine learning, terutama yang sensitif terhadap skala data, seperti regresi dan algoritma berbasis jarak.

* **Pengaruh Skala**: Setiap kolom dalam hasil scaling menunjukkan nilai yang telah disesuaikan, misalnya, nilai negatif menunjukkan bahwa nilai asli di bawah rata-rata, sedangkan nilai positif menunjukkan di atas rata-rata.

* **Contoh Nilai**: Pada output, nilai pertama kolom Production-based CO2 emissions adalah -0.82, yang berarti data asli berada jauh di bawah rata-rata dari seluruh dataset.

* **Persiapan untuk Model**: Dataset yang sudah di-scale ini siap digunakan untuk pelatihan model machine learning, meningkatkan akurasi dan performa model.

```python
# Split data menjadi data latih dan data uji
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,  # fitur yang sudah di-scaling
    y,         # target variabel
    test_size=0.2,  # 20% data akan digunakan untuk pengujian, 80% untuk pelatihan
    random_state=42  # random state untuk memastikan hasil yang konsisten setiap kali dijalankan
)

print("Data telah dibagi ke dalam data latih dan data uji:")
print(f"Ukuran X_train: {X_train.shape}")
print(f"Ukuran X_test: {X_test.shape}")
print(f"Ukuran y_train: {y_train.shape}")
print(f"Ukuran y_test: {y_test.shape}")
```

* **Pembagian Data**: Dataset telah dibagi menjadi dua subset, yaitu data latih (training) dan data uji (testing) untuk keperluan pelatihan dan evaluasi model.

* **Proporsi Pembagian**: Dengan menggunakan proporsi 80% untuk data latih dan 20% untuk data uji, jumlah data latih adalah 204 sampel, sedangkan jumlah data uji adalah 52 sampel.

* **Ukuran Fitur**: Baik X_train maupun X_test memiliki 5 fitur, yang menunjukkan bahwa model akan mempelajari dan menguji variabel dengan jumlah fitur yang sama.

* **Konsistensi**: Pengaturan random_state=42 memastikan bahwa pembagian data akan konsisten setiap kali kode dijalankan, sehingga hasil model dapat direproduksi.

* **Kesiapan untuk Pelatihan**: Dengan ukuran data latih yang memadai (204 sampel), model sudah siap untuk dilatih, yang dapat membantu meningkatkan akurasi dalam prediksi pada data uji.

```python
# train model Linear Regresion
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# train model Decision Tree Regressor
dt_model = DecisionTreeRegressor(random_state=42)
dt_model.fit(X_train, y_train)
```

```python
# train model Random Forest Regressor
rf_model = RandomForestRegressor(random_state=42)
rf_model.fit(X_train, y_train)
```

```python
# Predict on test data
y_pred_lr = model.predict(X_test)
y_pred_dt = dt_model.predict(X_test)
y_pred_rf = rf_model.predict(X_test)

# Evaluate model
mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)

mse_dt = mean_squared_error(y_test, y_pred_dt)
rmse_dt = np.sqrt(mse_dt)
r2_dt = r2_score(y_test, y_pred_dt)

mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)
r2_rf = r2_score(y_test, y_pred_rf)

# Print evaluation metrics
print("Linear Regression:")
print(f"MSE: {mse_lr:.2f}")
print(f"RMSE: {rmse_lr:.2f}")
print(f"R-squared: {r2_lr:.2f}")

print("\nDecision Tree:")
print(f"MSE: {mse_dt:.2f}")
print(f"RMSE: {rmse_dt:.2f}")
print(f"R-squared: {r2_dt:.2f}")

print("\nRandom Forest:")
print(f"MSE: {mse_rf:.2f}")
print(f"RMSE: {rmse_rf:.2f}")
print(f"R-squared: {r2_rf:.2f}")
```

* **Linear Regression**: MSE = 0.05, RMSE = 0.23, R-squared = 0.77 menunjukkan bahwa model memiliki kesalahan prediksi yang relatif rendah, dengan 77% variabilitas dalam data target yang dapat dijelaskan oleh model.

* **Decision Tree**: MSE = 0.02, RMSE = 0.14, R-squared = 0.91 menunjukkan performa yang lebih baik, dengan kesalahan prediksi yang lebih kecil dan 91% variabilitas yang dapat dijelaskan.

* **Random Forest**: MSE = 0.01, RMSE = 0.10, R-squared = 0.95 menunjukkan performa terbaik dengan kesalahan terendah dan kemampuan untuk menjelaskan 95% variabilitas dalam data target, menjadikannya model yang paling efisien di antara ketiga model yang diuji.

## **8. Hyperparameter Tuning dan Bootstraping**

```python
# Definisikan parameter yang ingin di-tune
param_grid = {
    'n_estimators': [50, 100, 200],  # Jumlah pohon dalam forest
    'max_depth': [None, 5, 10],  # Kedalaman maksimum pohon
    'min_samples_split': [2, 5, 10],  # Jumlah sampel minimum untuk membagi node
    'bootstrap': [True, False]  # Apakah menggunakan bootstrapping atau tidak
}

# Inisialisasi model Random Forest
rf_model = RandomForestRegressor(random_state=42)

# Inisialisasi GridSearchCV dengan model dan parameter yang ingin di-tune
grid_search = GridSearchCV(estimator=rf_model, param_grid=param_grid,
                          cv=5, scoring='neg_mean_squared_error', n_jobs=-1)

# Lakukan hyperparameter tuning dengan data latih
grid_search.fit(X_train, y_train)

# Cetak hasil terbaik
print("Best parameters found: ", grid_search.best_params_)
print("Lowest RMSE found: ", np.sqrt(-grid_search.best_score_))
```

* **Performansi Model**: Model mencapai RMSE terendah sebesar 0.10, menunjukkan akurasi yang baik dalam prediksi.

### **Interpretasi**

* **Kesalahan Kecil**: MSE sebesar 0.01 menunjukkan bahwa kesalahan rata-rata kuadrat antara prediksi dan nilai aktual sangat kecil.
* **Koefisien Determinasi Tinggi**: R-squared sebesar 0.96 menunjukkan bahwa 96% variasi dalam data target dapat dijelaskan oleh model, menandakan model sangat baik dalam menangkap pola data.

### **Alasan**

MSE (Mean Squared Error):

* Pengukuran Kesalahan: MSE mengukur rata-rata kuadrat kesalahan antara prediksi dan nilai aktual, memberikan gambaran tentang seberapa jauh prediksi menyimpang dari kebenaran.
* Sensitivitas terhadap Outlier: Metrik ini sangat sensitif terhadap outlier, sehingga dapat memberikan perhatian lebih pada kesalahan besar.

RMSE (Root Mean Squared Error):

* Interpretasi yang Mudah: RMSE memberikan nilai dalam satuan yang sama dengan data target, memudahkan interpretasi dan pemahaman kesalahan prediksi.
* Perbandingan Langsung: RMSE memungkinkan perbandingan yang lebih langsung antara model yang berbeda dalam hal performansi.

R-squared (Koefisien Determinasi):

* Proporsi Varians: R-squared menunjukkan proporsi variasi dalam data target yang dapat dijelaskan oleh model, memberikan gambaran tentang kekuatan hubungan antara variabel.
* Evaluasi Kualitas Model: Nilai R-squared mendekati 1 menunjukkan model yang baik, sedangkan nilai yang rendah menunjukkan model kurang efektif.


Secara keseluruhan, pemilihan metrik ini membantu dalam memberikan gambaran menyeluruh tentang performansi model, baik dari segi kesalahan prediksi maupun kualitas penjelasan model terhadap data.
